```python
# PROGRAM INTERAKTIF SEMUA LATIHAN

# fungsi-fungsi

def kategori_usia():
    umur = int(input("Masukkan usia: "))
    if 0 <= umur <= 12:
        print("Kategori: Anak-anak")
    elif umur <= 17:
        print("Kategori: Remaja")
    elif umur <= 59:
        print("Kategori: Dewasa")
    elif umur >= 60:
        print("Kategori: Lansia")
    else:
        print("Usia tidak valid")


def harga_tiket():
    def harga(umur):
        if umur <= 12:
            return 10000
        elif umur <= 17:
            return 15000
        elif umur <= 59:
            return 25000
        else:
            return 0

    t1 = int(input("Umur tiket 1: "))
    t2 = int(input("Umur tiket 2: "))
    t3 = int(input("Umur tiket 3: "))

    total = harga(t1) + harga(t2) + harga(t3)
    print("Total harga: Rp", total)


def diskon_restoran():
    umur = int(input("Masukkan umur: "))
    harga = int(input("Masukkan harga: "))

    if umur <= 12:
        diskon = 0.5
    elif umur >= 60:
        diskon = 0.3
    else:
        diskon = 0

    total = harga - (harga * diskon)
    print("Total bayar: Rp", total)


def beasiswa():
    nilai = int(input("Nilai rapor: "))
    penghasilan = int(input("Penghasilan orang tua: "))

    if nilai >= 90 and penghasilan < 5000000:
        print("Beasiswa Penuh")
    elif nilai >= 85 and penghasilan < 8000000:
        print("Beasiswa 50%")
    else:
        print("Tidak dapat beasiswa")


def akses_film():
    umur = int(input("Masukkan umur: "))
    kategori = input("Kategori (semua/remaja/dewasa): ").lower()

    if kategori == "semua":
        print("Boleh menonton")
    elif kategori == "remaja" and umur >= 13:
        print("Boleh menonton")
    elif kategori == "dewasa" and umur >= 18:
        print("Boleh menonton")
    else:
        print("Tidak boleh menonton")


def diskon_belanja():
    total = int(input("Total belanja: "))

    if total >= 500000:
        diskon = 0.2
    elif total >= 250000:
        diskon = 0.1
    else:
        diskon = 0

    bayar = total - (total * diskon)
    print("Total bayar: Rp", bayar)


def parkir():
    jenis = input("Jenis kendaraan (mobil/motor): ").lower()
    jam = int(input("Lama parkir (jam): "))

    if jenis == "mobil":
        biaya = 5000 * jam
    elif jenis == "motor":
        biaya = 2000 * jam
    else:
        biaya = 0

    if jam > 5:
        biaya += 10000

    print("Total biaya: Rp", biaya)


def transportasi():
    jarak = int(input("Jarak (km): "))

    if jarak <= 5:
        tarif = 5000
    elif jarak <= 10:
        tarif = 4000
    else:
        tarif = 3000

    total = jarak * tarif
    print("Total tarif: Rp", total)


def seleksi_programmer():
    coding = int(input("Nilai coding: "))
    interview = input("Nilai interview (A/B/dll): ").upper()

    if coding > 80:
        hasil_coding = "LOLOS"
    elif coding >= 60:
        hasil_coding = "DIPERTIMBANGKAN"
    else:
        hasil_coding = "GAGAL"

    if interview in ["A", "B"]:
        hasil_interview = "LOLOS"
    else:
        hasil_interview = "GAGAL"

    if hasil_coding in ["LOLOS", "DIPERTIMBANGKAN"] and hasil_interview == "LOLOS":
        print("Selamat Kamu Berhasil")
    else:
        print("Maaf Kamu Belum Berhasil")


def nomor_punggung():
    nomor = int(input("Nomor punggung: "))

    if nomor % 2 == 0:
        print("Target Attacker")
        if 50 <= nomor <= 100:
            print("Bisa jadi Captain")

    else:
        print("Defender")
        if nomor > 90:
            print("Playmaker")
        if nomor % 3 == 0 and nomor % 5 == 0:
            print("Keeper")


# MENU UTAMA
while True:
    print("\n=== MENU PROGRAM ===")
    print("1. Kategori Usia")
    print("2. Harga Tiket")
    print("3. Diskon Restoran")
    print("4. Beasiswa")
    print("5. Akses Film")
    print("6. Diskon Belanja")
    print("7. Parkir")
    print("8. Transportasi")
    print("9. Seleksi Programmer")
    print("10. Nomor Punggung")
    print("0. Keluar")

    pilihan = input("Pilih menu: ")

    if pilihan == "1":
        kategori_usia()
    elif pilihan == "2":
        harga_tiket()
    elif pilihan == "3":
        diskon_restoran()
    elif pilihan == "4":
        beasiswa()
    elif pilihan == "5":
        akses_film()
    elif pilihan == "6":
        diskon_belanja()
    elif pilihan == "7":
        parkir()
    elif pilihan == "8":
        transportasi()
    elif pilihan == "9":
        seleksi_programmer()
    elif pilihan == "10":
        nomor_punggung()
    elif pilihan == "0":
        print("Terima kasih!")
        break
    else:
        print("Pilihan tidak valid")
```
