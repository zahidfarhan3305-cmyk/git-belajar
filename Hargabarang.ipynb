```python

daftar_promo = ["HEMAT10", "PROMO20", "SALE30"]
data_cache = {"id": 101, "nama": "Budi"} 
biaya_admin = 2500
status_sistem = 6     
masker_izin = 3       


print("=== SISTEM ANALISIS PENJUALAN TOKO ONLINE ===")
nama_pembeli = input("Nama Pelanggan: ")
jumlah_barang = int(input("Jumlah barang: "))
harga_satuan = int(input("Harga per barang: "))
is_member = input("Status member (y/n): ").lower() == 'y'
kode_input = input("Masukkan kode diskon: ")


total = jumlah_barang * harga_satuan
if jumlah_barang > 10:
    potongan_borong = total * 0.05
    total = total - potongan_borong


total += biaya_admin


bonus_poin = 0
if is_member and total > 500000:
    bonus_poin = 100

if kode_input in daftar_promo or (is_member and kode_input == "MEMBERBARU"):
    print(f"-> Kode '{kode_input}' AKTIF! Potongan Rp10.000 diterapkan.")
    total -= 10000
else:
    print("-> Kode diskon tidak valid atau tidak tersedia.")


if total >= 1000000:
    kategori = "Pembelian Besar"
elif total >= 500000:
    kategori = "Pembelian Sedang"
else:
    kategori = "Pembelian Kecil"

data_pelanggan = {"id": 101, "nama": "Budi"} 

cek_identitas = data_pelanggan is data_cache 


bw_and = status_sistem & masker_izin   # Irisan izin
bw_or  = status_sistem | masker_izin   # Gabungan izin
bw_xor = status_sistem ^ masker_izin   # Perbedaan izin
bw_not = ~status_sistem                # Inversi bit
bw_lshift = status_sistem << 1         # Geser kiri (x2)
bw_rshift = status_sistem >> 1         # Geser kanan (/2)


print("\n" + "="*30)
print(f"LAPORAN TRANSAKSI: {nama_pembeli}")
print(f"Total Bayar    : Rp{total:,.0f}")
print(f"Kategori Belanja: {kategori}")
print(f"Bonus Loyalitas : {bonus_poin} Poin")
print(f"Status Cache    : {'Identik' if cek_identitas else 'Objek Baru'}")
print("ANALISIS BITWISE (SISTEM KEAMANAN):")
print(f"AND: {bw_and} | OR: {bw_or} | XOR: {bw_xor}")
print(f"NOT: {bw_not} | L-Shift: {bw_lshift} | R-Shift: {bw_rshift}")
print("="*30)

```
