```python

daftar_mahasiswa = ["Andi", "Budi", "Citra", "Dina"]
print("=== SISTEM EVALUASI NILAI LENGKAP ===")
nama_input = input("Masukkan Nama Mahasiswa: ")

if nama_input in daftar_mahasiswa:
    
    tugas = float(input("Masukkan Nilai Tugas: "))
    uts = float(input("Masukkan Nilai UTS: "))
    uas = float(input("Masukkan Nilai UAS: "))
    
    

    
    nilai_akhir = (tugas * 0.3) + (uts * 0.3) + (uas * 0.4)

    
    is_lulus = nilai_akhir >= 75
    is_sempurna = (nilai_akhir >= 75) and (uas >= 70)

    
    nilai_backup = nilai_akhir
    cek_identitas = nilai_akhir is nilai_backup

  
    n = int(nilai_akhir)
    m = 3 
    
    bw_and   = n & m   
    bw_or    = n | m  
    bw_xor   = n ^ m   
    bw_not   = ~n      
    bw_left  = n << 1  
    bw_right = n >> 1  

    # --- OUTPUT HASIL 
    print("\n" + "="*40)
    print(f"HASIL EVALUASI UNTUK: {nama_input.upper()}")
    print(f"Nilai Akhir (setelah bonus): {nilai_akhir:.2f}")
    print(f"Status Lulus: {is_lulus} | Sempurna: {is_sempurna}")
    print(f"Cek Identitas Variabel: {cek_identitas}")
    
    print("-" * 15 + " OPERASI BITWISE " + "-" * 15)
    print(f"Angka Dasar (n)  : {n} (Biner: {bin(n)})")
    print(f"Angka Pembanding : {m} (Biner: {bin(m)})")
    print(f"1. AND (&)       : {bw_and}")
    print(f"2. OR (|)        : {bw_or}")
    print(f"3. XOR (^)       : {bw_xor}")
    print(f"4. NOT (~)       : {bw_not}")
    print(f"5. Shift Left (<<): {bw_left}")
    print(f"6. Shift Right(>>): {bw_right}")
    print("="*40)

else:
    print(f"Maaf, nama '{nama_input}' tidak terdaftar.")
```
