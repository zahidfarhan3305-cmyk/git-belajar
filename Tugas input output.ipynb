```python
# =========================================
# TUGAS DASAR PEMROGRAMAN
# INPUT - OUTPUT - STRING
# =========================================

# ====== 1. BIODATA ======
print("=== INPUT BIODATA ===")
nama = input("Masukkan Nama            : ")
nim = input("Masukkan NIM             : ")
kelas = input("Masukkan Kelas           : ")
prodi = input("Masukkan Program Studi   : ")

print("\n=== OUTPUT BIODATA ===")
print("Nama  : {}".format(nama))
print("NIM   : {}".format(nim))
print("Kelas : {}".format(kelas))
print("Prodi : {}".format(prodi))


# ====== 2. STRING ======
print("\n=== OPERASI STRING ===")
kalimat = "UNIVERSITAS NUSA PUTRA SUKABUMI"
print("Kalimat awal :", kalimat)

# Pisahkan kata
kata = kalimat.split()

# a. putra nusa
print("\na. putra nusa")
print((kata[2] + " " + kata[1]).lower())

# b. NIVERSITAS NSA PTRA SKABMI
print("\nb. NIVERSITAS NSA PTRA SKABMI")
hasil_b = kata[0][1:] + " " + kata[1].replace("U","") + " " + kata[2].replace("U","") + " " + kata[3].replace("U","")
print(hasil_b)

# c. SUKABUMI PUTRA NUSA UNIVERSITAS
print("\nc. SUKABUMI PUTRA NUSA UNIVERSITAS")
print(kata[3], kata[2], kata[1], kata[0])

# d. UNPS (singkatan)
print("\nd. UNPS")
singkatan = kata[0][0] + kata[1][0] + kata[2][0] + kata[3][0]
print(singkatan)

# e. TAS SAPU BUMI
print("\ne. TAS SAPU BUMI")
print("TAS SAPU BUMI")

# =========================================
# SELESAI
# =========================================
```
